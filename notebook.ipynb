````python
#Q2:

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("./data/land_mines.csv")

classification_X_raw = df[["voltage", "height", "soil"]]
classification_y = df["mine_type"]
(classification_X_train_raw, classification_X_valid_raw, classification_y_train, classification_y_valid) = train_test_split(
    classification_X_raw,
    classification_y,
    test_size=0.5,
    random_state=65,
)

classification_scaler = MinMaxScaler()
classification_X_train_scaled = classification_scaler.fit_transform(classification_X_train_raw)
classification_X_valid_scaled = classification_scaler.transform(classification_X_valid_raw)

classification_k = 11
classification_model = KNeighborsClassifier(n_neighbors=classification_k).fit(classification_X_train_scaled, classification_y_train)
classification_valid_hat = classification_model.predict(classification_X_valid_scaled)

confusion_matrix = pd.crosstab(
    classification_y_valid.to_numpy(),
    classification_valid_hat,
    rownames=["Actual"],
    colnames=["Predicted"],
)
print(confusion_matrix)


````

### *AI-assisted solution in full
After the evaluation of AI suggestions, incorporate the accepted revisions into your Phase 1 solution and provide the final Phase 2 solution in full in the following cells.

````python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score

df = pd.read_csv("./data/land_mines.csv")

classification_X_raw = df[["voltage", "height", "soil"]]
classification_y = df["mine_type"]

(classification_X_train_raw, classification_X_valid_raw,
 classification_y_train, classification_y_valid) = train_test_split(
    classification_X_raw,
    classification_y,
    test_size=0.5,
    random_state=65,
    stratify=classification_y,   # keep class proportions equal in both halves
)

# Sanity check: class proportions should match across the two sets
print(classification_y_train.value_counts(normalize=True).sort_index())
print(classification_y_valid.value_counts(normalize=True).sort_index())

classification_scaler = MinMaxScaler()
classification_X_train_scaled = classification_scaler.fit_transform(classification_X_train_raw)
classification_X_valid_scaled = classification_scaler.transform(classification_X_valid_raw)

classification_k = 11
classification_model = KNeighborsClassifier(n_neighbors=classification_k).fit(classification_X_train_scaled, classification_y_train)
classification_valid_hat = classification_model.predict(classification_X_valid_scaled)

labels = np.sort(classification_y.unique())   # keep every class, even if never predicted

confusion_matrix = pd.crosstab(
    classification_y_valid.to_numpy(),
    classification_valid_hat,
    rownames=["Actual"],
    colnames=["Predicted"],
).reindex(index=labels, columns=labels, fill_value=0)

print(confusion_matrix)
print(f"\nOverall accuracy: {accuracy_score(classification_y_valid, classification_valid_hat):.3f}")

# Per-class performance
diag = np.diag(confusion_matrix.to_numpy())
per_class = pd.DataFrame({
    "n_actual": confusion_matrix.sum(axis=1),
    "recall (row-wise)": diag / confusion_matrix.sum(axis=1).replace(0, np.nan),
    "precision (col-wise)": diag / confusion_matrix.sum(axis=0).replace(0, np.nan),
}, index=labels)
print("\n", per_class.round(3))

# Most common mistakes
errors = confusion_matrix.stack()
errors = errors[[a != p for a, p in errors.index]].sort_values(ascending=False)
print("\nTop confusions (actual, predicted):")
print(errors.head(5))
````
