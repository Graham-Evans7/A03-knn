#3:
I chose k = 11 following the suggestion learned in class of k = sqrt(n), then trying similar values to find accuracy. After splitting up the data, n = 169 and therefore k = 13. However, after trying values close to 13, I found that 11 had a slightly higher accuracy than the others, so I chose to stick with that.
#4:
The algorithm is correct around 44% of the time. Specifically, it is very accurate for mine type 2, but less accurate with mine types 1 and 4, and not very accurate with mine types 3 and 5.
#5:
Given the inconsistency in accuracy for different mine types (e.g. 2 vs. 5), it is important to take those inconsistencies into consideration. For example, if the estimated mine type is 2, it is probably correct. However, a guess of mine type 5 is probably incorrect. More generally, look at the accuracy of each mine type predictiona and use judgement based off of it. It shouldn't be the final decision when choosing a mine type but more of a basis to work off of.


```python
#Q2:

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("./data/land_mines.csv")

classification_X_raw = df[["voltage", "height", "soil"]]
classification_y = df["mine_type"]
(classification_X_train_raw, classification_X_valid_raw, classification_y_train, classification_y_valid) = train_test_split(
    classification_X_raw,
    classification_y,
    test_size=0.5,
    random_state=65,
)

classification_scaler = MinMaxScaler()
classification_X_train_scaled = classification_scaler.fit_transform(classification_X_train_raw)
classification_X_valid_scaled = classification_scaler.transform(classification_X_valid_raw)

classification_k = 11
classification_model = KNeighborsClassifier(n_neighbors=classification_k).fit(classification_X_train_scaled, classification_y_train)
classification_valid_hat = classification_model.predict(classification_X_valid_scaled)

confusion_matrix = pd.crosstab(
    classification_y_valid.to_numpy(),
    classification_valid_hat,
    rownames=["Actual"],
    colnames=["Predicted"],
)
print(confusion_matrix)


```
